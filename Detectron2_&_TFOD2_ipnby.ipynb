**Question 1. What is Detectron2 and how does it differ from previous object detection frameworks?**

**Answer.** Detectron2 is Facebook AI Research’s next-generation library for object detection and segmentation, built on PyTorch. It replaces the older Detectron, which was based on Caffe2. The big difference is flexibility and speed: Detectron2 has a fully modular design, letting you plug in new backbones, heads, and training configs without wrestling with messy code. It supports state-of-the-art models like Faster R-CNN, Mask R-CNN, RetinaNet, DensePose, and panoptic segmentation out of the box. Because it uses PyTorch, debugging is easier and training is faster on modern hardware. It also integrates better with current deep-learning workflows, provides cleaner APIs, and scales from research experiments to production. Overall, Detectron2 is more efficient, developer-friendly, and extensible than previous frameworks.

**Question 2: Explain the process and importance of data annotation when working with
Detectron2.**

**Answer.** Data annotation is the backbone of any Detectron2 workflow because the model learns directly from the labeled examples you provide. The process starts with collecting images and marking every object using tools like LabelMe, CVAT, or VIA. Each object must be assigned a class label and a precise bounding box, polygon, or mask depending on the task. After annotation, the data is converted to Detectron2’s preferred COCO-style JSON format, which includes image metadata, categories, and annotations. This structured labeling ensures the model understands what to detect, how objects look, and where they appear. High-quality annotations reduce noise, improve training stability, and directly boost model accuracy. Poor or inconsistent labeling, on the other hand, leads to unreliable predictions and wasted training time.

**Question 3: Describe the steps involved in training a custom object detection model
 using Detectron2.**

**Answer.** Training a custom object detection model in Detectron2 follows a clear sequence. First, prepare and annotate your dataset, then convert it to COCO-style JSON. Next, register the dataset with Detectron2 using DatasetCatalog and MetadataCatalog. After that, choose a suitable config file (e.g., Faster R-CNN or Mask R-CNN) and modify key settings like number of classes, dataset paths, learning rate, and batch size. Load pretrained weights to speed up convergence. Then start training using the built-in DefaultTrainer, which handles the training loop, evaluation, and checkpointing. Once training finishes, test the model on validation images and measure metrics like mAP. Finally, fine-tune hyperparameters or improve your dataset if performance is weak, then export the trained model for inference.

**Question 4: What are evaluation curves in Detectron2, and how are metrics like mAP
and IoU interpreted?**

**Answer.**Evaluation curves in Detectron2 visualize how well a model performs across different confidence thresholds or IoU levels. They help you see patterns like over-prediction, under-prediction, or inconsistent bounding boxes. Two key metrics dominate: IoU (Intersection over Union) and mAP (Mean Average Precision). IoU measures how accurately a predicted box overlaps with the ground-truth box; higher IoU means tighter, cleaner localization. mAP summarizes both precision and recall across multiple IoU thresholds, giving a single score for overall detection quality. A high mAP indicates the model is identifying objects correctly and consistently across classes. If IoU is high but mAP is low, class confusion or inconsistent confidence scores are usually the real problem, not box placement.

**Question 5: Compare Detectron2 and TFOD2 in terms of features, performance, and
ease of use.**

**Answer.** Detectron2 and TFOD2 both handle object detection well, but they differ in workflow and flexibility. Detectron2, built on PyTorch, is more modular and research-friendly. It supports advanced models like Mask R-CNN, DensePose, and panoptic segmentation with cleaner APIs and faster debugging. Performance is strong, especially for segmentation-heavy tasks. TFOD2, built on TensorFlow, is more production-oriented and easier for beginners due to prebuilt pipelines, TFRecords, and TensorFlow Serving integration. It offers a wide variety of pretrained models and simpler deployment on mobile or TensorFlow Lite. However, customizing architectures in TFOD2 is more restrictive. In short: Detectron2 wins in flexibility and cutting-edge research, while TFOD2 is smoother for straightforward training and large-scale deployment.

**Question 8. Write a script to download pretrained weights and configure paths for training in Detectron2.**

**Answer.**

```python
import os
import requests
from detectron2.config import get_cfg
from detectron2 import model_zoo

def download_file(url, save_path):
    """Download file with streaming to avoid memory issues."""
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    print(f"Downloading: {url}")

    response = requests.get(url, stream=True)
    if response.status_code != 200:
        raise Exception(f"Failed to download file: {url}")

    with open(save_path, "wb") as f:
        for chunk in response.iter_content(chunk_size=1024):
            if chunk:
                f.write(chunk)
    print(f"Saved to: {save_path}")


def setup_detectron2_config(
    model_zoo_config="COCO-InstanceSegmentation/mask_rcnn_R_50_FPN_3x.yaml",
    weights_directory="weights/",
    dataset_train="my_dataset_train",
    dataset_test="my_dataset_val"
):
    cfg = get_cfg()


    config_path = model_zoo.get_config_file(model_zoo_config)
    cfg.merge_from_file(config_path)


    weights_url = model_zoo.get_checkpoint_url(model_zoo_config)

    weights_path = os.path.join(weights_directory,
                                os.path.basename(weights_url))

    if not os.path.exists(weights_path):
        download_file(weights_url, weights_path)
    else:
        print(f"Pretrained weights already exist: {weights_path}")


    cfg.MODEL.WEIGHTS = weights_path


    cfg.DATASETS.TRAIN = (dataset_train,)
    cfg.DATASETS.TEST = (dataset_test,)


    cfg.OUTPUT_DIR = "./output/"
    os.makedirs(cfg.OUTPUT_DIR, exist_ok=True)


    cfg.SOLVER.IMS_PER_BATCH = 2
    cfg.SOLVER.BASE_LR = 0.00025
    cfg.SOLVER.MAX_ITER = 3000
    cfg.MODEL.ROI_HEADS.BATCH_SIZE_PER_IMAGE = 128
    cfg.MODEL.ROI_HEADS.NUM_CLASSES = 1  # change if more classes

    print("\nConfiguration Completed!")
    print(f"Model config: {model_zoo_config}")
    print(f"Weights saved at: {weights_path}")
    print(f"Output dir: {cfg.OUTPUT_DIR}")

    return cfg


if __name__ == "__main__":
    cfg = setup_detectron2_config()
```

**Question 9. Show the steps and code to run inference using a trained Detectron2 model on a new image.**

**Answer.**

```python
import cv2
import torch
from detectron2.config import get_cfg
from detectron2.engine import DefaultPredictor
from detectron2 import model_zoo
from detectron2.utils.visualizer import Visualizer
from detectron2.data import MetadataCatalog


def load_model(config_path, weights_path, score_threshold=0.5):
    cfg = get_cfg()


    cfg.merge_from_file(config_path)


    cfg.MODEL.WEIGHTS = weights_path


    cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = score_threshold


    predictor = DefaultPredictor(cfg)
    return predictor, cfg



def run_inference(predictor, cfg, image_path, output_path="result.jpg"):
    img = cv2.imread(image_path)

    outputs = predictor(img)


    metadata = MetadataCatalog.get(cfg.DATASETS.TRAIN[0])
    v = Visualizer(img[:, :, ::-1], metadata=metadata, scale=1.0)
    result = v.draw_instance_predictions(outputs["instances"].to("cpu"))

    final_img = result.get_image()[:, :, ::-1]
    cv2.imwrite(output_path, final_img)

    print(f"Inference completed! Result saved at: {output_path}")


if __name__ == "__main__":
    config_file = "output/config.yaml"
    weights_file = "output/model_final.pth"

    predictor, cfg = load_model(config_file, weights_file)

    input_image = "test.jpg"

    run_inference(predictor, cfg, input_image, "prediction.jpg")
```
